## Download Configuration File via SDK

We assume that you have already saved the token as described in the Quick Start guide.

```{eval-rst}

.. seealso::
    The steps to get and save token can be found in `Quick Start <../quick_start/index.html#save-token>`_.

```

Via SDK, you can download the configuration file for the machine of the QC-Setup using the QC-Setup ID or QPU ID,
which we have given IDs and the code examples in the modal of **Get Port Number and Configuration File**.

```{eval-rst}

.. seealso::
    The steps to get and save token can be found in `Quick Start <../quick_start/index.html#save-token>`_.

```

The dropdown menu allows you to select either the QC-Setup ID or the QPU ID to search the configuration file for the machine of the QC-Setup.

### By QC-Setup ID

By QC-Setup ID, you can search for the configuration file of the machine associated with the selected QC-Setup.

![Select QC-Setup ID](./demo_qcstup.png)

````python
from qctss_client import QCTSSClient

client = QCTSSClient()

configs_result = client.get_config_by_qcsetup(8)
configs_result
````

### By QPU ID

In some cases, the QC-Setup ID may not be available, then you can search for the configuration file from other machine using the same chip by QPU Id.

![Select QPU ID](./demo_qpu.png)

````python
from qctss_client import QCTSSClient

client = QCTSSClient()

configs_result = client.get_config_by_qpu(1)
configs_result
````

```{eval-rst}

.. seealso::
    The return object of :func:`qctss_client.client.core.QCTSSClient.get_config_by_qcsetup` 
    and :func:`qctss_client.client.core.QCTSSClient.get_config_by_qpu` are 
    of type `ConfigurationsResult`. 
    For more details, see :class:`qctss_client.models.configuration.ConfigurationsResult`.

```

### Save Configuration

Now we pick the configuration we want to save.

````python
selected_config = configs_result.get_configuration(1)
selected_config
````

```{eval-rst}

By the function :func:`qctss_client.models.configuration.ConfigurationsResult.save`, 
the configuration files will be saved in the new created directory specified by the file path.
```

````python
configs_result[0].save("./.quamstate")
````
